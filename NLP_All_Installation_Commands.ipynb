# NLP Assignments — Complete Installation Guide

This notebook collects the installation commands for the Python libraries used across the NLP assignments, including NLTK, spaCy, Transformers/BERT, PyTorch, scikit-learn, Gensim, Stanza, and plotting/network libraries.

**Important:** Run the operating-system setup commands in a terminal, not in a Python cell. Run the Python cells below inside Jupyter. Internet access is needed to install packages and download language/model data.

## 1. Ubuntu / WSL setup (run in Terminal)

Create and activate a virtual environment, then install Jupyter inside it.

```python
sudo apt update
sudo apt install -y python3-pip python3-venv
python3 -m venv nlp-env
source nlp-env/bin/activate
python -m pip install --upgrade pip
python -m pip install jupyter ipykernel
python -m ipykernel install --user --name nlp-env --display-name "Python (nlp-env)"
jupyter notebook
```

## 8. Quick notes / troubleshooting

- **Choose the correct kernel:** In Jupyter, select **Python (nlp-env)** if you created the virtual environment above.
- **Ubuntu externally-managed-environment error:** Install packages inside the virtual environment; avoid `sudo pip install`.
- **Windows PowerShell activation blocked:** Run `Set-ExecutionPolicy -Scope Process -ExecutionPolicy Bypass` in that PowerShell window, then activate the environment again.
- **PyTorch `c10.dll` / `WinError 1114` on Windows:** This is usually an environment or binary compatibility issue. Try a fresh virtual environment and install PyTorch using the official selector at https://pytorch.org/get-started/locally/ rather than changing an existing TensorFlow environment.
- **Model downloads:** Transformers models and NLTK/spaCy/Stanza resources may require internet access the first time.
- **If you only need a subset of assignments:** You can install just the libraries used by those assignments instead of every package listed here.
- These commands install packages; they do not include the assignment implementations themselves.
