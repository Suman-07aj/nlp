```python
import string
import re , math, random
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer , WordNetLemmatizer
from nltk.util import ngrams

nltk.download("punkt",quiet=True)
nltk.download("stopwords",quiet=True)
nltk.download("wordnet",quiet=True)

po="""Language is a method of communication with the
help of which we can speak, read and write :) https://test.com.
Natural Language Processing (NLP) is the sub field of computer
science especially Artificial Intelligence (AI)
that is concerned about enabling computers to understand and
process human language. We have
various open-source NLP tools but NLTK (Natural Language Toolkit)
scores very high when it comes
to the ease of use and explanation of the concept. The learning
curve of Python is very fast and NLTK
is written in Python so NLTK is also having very good learning
kit. NLTK has incorporated most of
the tasks like tokenization, stemming, Lemmatization, Punctuation,
Character Count, and Word count.
It is very elegant and easy to work with."""

# Remove URLs
p = re.sub(r"https?://\S+|www\.\S+", " ", po)

# Tokenize and convert into lowercase
tokens = word_tokenize(p.lower())

# Remove punctuation
toc_no_pun = [token for token in tokens if token not in string.punctuation]

# Remove stopwords
stop_words = set(stopwords.words('english'))
token_no_stopw = [token for token in toc_no_pun if token not in stop_words]

# Remove duplicates and sort
base_clean_list = sorted(list(set(token_no_stopw)))

# Print
print("--- Base Cleaned List ---")
print(base_clean_list)
print("\n" + "=" * 50 + "\n")
```

```python
#porter stemmer

porter=PorterStemmer()

porter_stems=sorted(list(set([porter.stem(word) for word in base_clean_list])))

print("---PORTER STEMS ---")
print(porter_stems)
print("\n" + "=" * 50 + "\n")


lemmatizer=WordNetLemmatizer()

lematization=sorted(list(set(lemmatizer.lemmatize(word) for word in base_clean_list)))

print('------LEMMATIZED WORDS-----')
print(lematization)
print('\n','='*50,'\n')


# Bigrams

ngram = list(ngrams(base_clean_list, 2))

for gram in ngram:
    print(" ".join(gram))

print('\nTotal bigram sequences generated:', len(ngram))
```

```python

# Bigram Language Model

from collections import Counter

# Training data
train = [word_tokenize(p.lower())]

unigram, bigram = Counter(), Counter()
vocab = {"<s>", "</s>", "<unk>"}

for sent in train:
    vocab.update(sent)
    seq = ["<s>"] + sent + ["</s>"]
    unigram.update(seq[:-1])
    bigram.update(zip(seq[:-1], seq[1:]))

V = len(vocab)

# Add-one smoothing
def prob(word, prev):
    return (bigram[(prev, word)] + 1) / (unigram[prev] + V)

# Sentence probability
def sentence_prob(words):
    seq = ["<s>"] + words + ["</s>"]
    p = 1
    for prev, word in zip(seq, seq[1:]):
        p *= prob(word if word in vocab else "<unk>", prev)
    return p

# Perplexity
def perplexity(words):
    seq = ["<s>"] + words + ["</s>"]
    logp = sum(math.log(prob(w if w in vocab else "<unk>", prev))
               for prev, w in zip(seq, seq[1:]))
    return math.exp(-logp / (len(seq) - 1))

test = word_tokenize("language is useful")
print("Probability:", sentence_prob(test))
print("Perplexity:", perplexity(test))

# Sampling
def sample(seed=42, length=10):
    rng, out, prev = random.Random(seed), [], "<s>"
    for _ in range(length):
        choices = sorted(vocab)
        w = rng.choices(choices, [prob(x, prev) for x in choices])[0]
        if w == "</s>":
            break
        if w != "<s>":
            out.append(w)
        prev = w
    return " ".join(out)

print("Sample:", sample())
```

```python
from nltk import CFG
from nltk.parse import ChartParser

grammar = CFG.fromstring('''
S -> NP VP
NP -> Det N | NP PP
VP -> V NP | VP PP
PP -> P NP
Det -> 'the'
N -> 'man' | 'telescope'
V -> 'saw'
P -> 'with'
''')

parser = ChartParser(grammar)

sentence = "the man saw the man with the telescope".split()

for tree in parser.parse(sentence):
    print(tree)
    print()
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp("The screen is bright. The camera is excellent.")
for t in doc:
    print(t.text, t.pos_, t.dep_, "head=", t.head.text)

pairs = set()
for t in doc:
    if t.dep_ == "amod":
        pairs.add((t.head.text, t.text))  # aspect, opinion
    if t.dep_ in {"acomp", "attr"}:
        for child in t.head.children:
            if child.dep_ == "nsubj":
                pairs.add((child.text, t.text))
print("Aspect-opinion:", pairs)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

docs = ["cats are animals", "language models predict words",
        "NLP processes human language"]
query = "language models"

vec=TfidfVectorizer()
X=vec.fit_transform(docs + [query])

score=cosine_similarity(X[-1],X[:-1]).ravel()
for i in score.argsort()[::-1][:2]:
    print(round(score[i],3),docs[i])
```

```python
corpus=[word_tokenize("cats chase small animals"),word_tokenize("dogs chase cats")]

uni, pairs=Counter(),Counter()
tot=0

for sent in corpus:
    uni.update(sent)
    for i, w in enumerate(sent):

        for j in range(max(0,i-2) , min(len(sent),i+3)):
            if i!=j:
                pairs[(w,sent[j])]+=1
                tot+=1
for (w,c) ,n in pairs.items():
    pmi=math.log2((tot*n ) / uni[w] *uni[c])
    print(w, c, "PPMI=", round(max(0,pmi), 3))
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp("John met Mary. He thanked her. Mary smiled.")
mentions = []
gender = {"john":"M", "he":"M", "him":"M", "mary":"F", "she":"F", "her":"F"}
for ent in doc.ents:
    if ent.label_ == "PERSON":
        mentions.append((ent.text, ent.start, gender.get(ent.text.lower(), "?"), "name"))
for t in doc:
    if t.pos_ == "PRON":
        mentions.append((t.text, t.i, gender.get(t.text.lower(), "?"), "pronoun"))
mentions.sort(key=lambda x: x[1])
print("Mentions:", mentions)

def score(antecedent, mention):
    # tuple: text, token_position, gender, kind
    if antecedent[1] >= mention[1]: return -999
    if antecedent[2] != "?" and mention[2] != "?" and antecedent[2] != mention[2]:
        return -999
    return (3 if antecedent[2] == mention[2] else 0) - (mention[1]-antecedent[1]) * 0.1

for m in mentions:
    if m[3] == "pronoun":
        prior = [a for a in mentions if a[1] < m[1]]
        if prior:
            best = max(prior, key=lambda a: score(a,m))
            print(m[0], "->", best[0])
```
